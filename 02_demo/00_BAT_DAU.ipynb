# Bắt đầu — Phòng thực nghiệm Posit cho AI

Các notebook đã kèm bảng số liệu và hình từ kết quả sẵn có. Bạn có thể đọc ngay mà chưa cần chạy huấn luyện.

## Cách thao tác

1. Đọc [Hướng dẫn notebook](HUONG_DAN_NOTEBOOK.md) để thiết lập kernel một lần.
2. Mở notebook, giữ `RUN_EXPERIMENT = False`, chọn **Run All** để đọc số liệu và xuất báo cáo.
3. Khi muốn chạy thực nghiệm, đổi biến thành `True` rồi chạy từ đầu. Dữ liệu cũ được sao lưu.
4. Nhấn **Ctrl+S / Cmd+S** để giữ output trong `.ipynb`.

## Đọc kết quả thế nào?

- 01: kiểm chứng định dạng. `mismatches = 0` là khớp trên các ca kiểm tra đã chạy; thiếu SoftPosit thì chưa đối chiếu phần đó.
- 02: mốc FP32 để so sánh, không nhầm với FP32 huấn luyện riêng ở D.
- A: phân bố giá trị và độ chính xác biểu diễn.
- B: độ chính xác sau giảm bit trọng số; xem cả scaling và sai số giữa seed.
- C: tác động cách cộng dồn; FP64 mô phỏng quire.
- D: khả năng học khi giảm bit; bảng cuối cùng và lịch sử từng epoch.
- E: sai số xấp xỉ hàm kích hoạt.
- 08: vẽ lại hình B/C/D khi JSON thay đổi.
- 09/10: ví dụ giải mã và hình lý thuyết.
- 11: tiện ích gộp hai phần kết quả D, có thể bỏ qua.
